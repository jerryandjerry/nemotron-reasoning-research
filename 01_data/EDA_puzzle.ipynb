```python
import pandas as pd

CATEGORY = 'equation_cryptarithm'  # change to: bit_manipulation, text_encryption, number_conversion, unit_conversion, gravitational_constant, equation_numeric, equation_cryptarithm

df = pd.read_csv(f'train_cot_{CATEGORY}.csv', dtype=str)
print(f"{CATEGORY}: {len(df)} rows, cols={df.columns.tolist()}")
df.head()
```

```python
# View by puzzle ID
PUZZLE_ID = '00457d26'  # change this

row = df[df['id'] == PUZZLE_ID]
if len(row) > 0:
    row = row.iloc[0]
    print(f"ID: {row['id']}")
    print(f"Answer: {repr(row['answer'])}")
    print()
    print("=== PROMPT ===")
    print(row['prompt'])
    print()
    print("=== SOLVER COT ===")
    cot = row['solver_cot']
    if pd.notna(cot) and cot:
        print(cot)
    else:
        print("(empty)")
else:
    print(f"ID {PUZZLE_ID} not found in {CATEGORY}")
```
